# Refresh partitions (point-and-click)

For support teams: no code editing needed.

1. Set the workspace and semantic model in the parameters cell and run it. The panel appears.
2. In the panel: pick a table, select partitions (row counts and last refresh times are shown), leave **Dry run** ticked and click **Plan**.
3. To refresh for real: **untick Dry run**, tick **I confirm**, then **run the last cell (`panel.run()`)**. That cell shows as *running* until the refresh finishes, prints a log and a "still running" line every minute, and shows the result at the end.

```python
%pip install fabric-partition-refresh
```

## Open the panel

```python
from fabric_partition_refresh import launch_refresh_ui

WORKSPACE = "My Workspace"
SEMANTIC_MODEL = "My Semantic Model"

panel = launch_refresh_ui(WORKSPACE, SEMANTIC_MODEL)
```

## Start the refresh
Run this cell **after** unticking *Dry run* and ticking *I confirm* in the panel above. It stays *running* until the refresh is done.

```python
panel.run()
```

## If you need the result again
The outcome is always stored on the panel:

```python
panel.last_report.verification if panel.last_report else panel.last_error
```
